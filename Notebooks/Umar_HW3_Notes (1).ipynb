1) Top 1% orders by total value

Proposition: Find the biggest orders by money.

Question: Which orders are in the top 1% by total price?

Answer: Add up each order’s line totals and keep only the top 1%.

How I’m solving it: Join Orders → OrderLines, sum UnitPrice*Quantity, sort high to low, use TOP (1) PERCENT.

Why special: Shows multi-join + aggregation and the less common TOP PERCENT.

```python
SELECT TOP (1) PERCENT
       o.OrderID,
       SUM(ol.UnitPrice * ol.Quantity) AS OrderTotal
FROM Sales.Orders AS o
JOIN Sales.OrderLines AS ol
  ON ol.OrderID = o.OrderID
GROUP BY o.OrderID
ORDER BY OrderTotal DESC;
```

4) Customers with no orders in 2012 (anti-join)

Proposition: Find customers who did not place orders during 2012.

Question: Which customers have zero orders in 2012?

Answer: Keep customers whose 2012 order match is NULL.

How I’m solving it: LEFT JOIN Customers → Orders (orders filtered to 2012 in ON), then WHERE o.OrderID IS NULL.

Why special: Correct anti-join pattern: filter the non-preserved table in ON, then IS NULL.

```python
SELECT
  c.CustomerID,
  c.CustomerName
FROM Sales.Customers AS c
LEFT JOIN Sales.Orders AS o
  ON  o.CustomerID = c.CustomerID
  AND o.OrderDate >= '2012-01-01'
  AND o.OrderDate <  '2013-01-01'
WHERE o.OrderID IS NULL
ORDER BY c.CustomerName;
```

9) Customers from the same city (self-join)

Proposition: Find pairs of customers who live in the same city.

Question: Which two customers share the same delivery city?

Answer: Join the Customers table to itself by city ID.

How I’m solving it: Self-join Customers as c1 and c2 on DeliveryCityID; use c1.CustomerID < c2.CustomerID to avoid duplicates; join Cities to show the name.

Why special: Shows a self-join and a clean technique to avoid duplicate pairs.

```python
SELECT
   c1.CustomerName AS Customer1,
   c2.CustomerName AS Customer2,
   ci.CityName
FROM Sales.Customers AS c1
JOIN Sales.Customers AS c2
   ON c1.DeliveryCityID = c2.DeliveryCityID
  AND c1.CustomerID < c2.CustomerID
JOIN Application.Cities AS ci
   ON ci.CityID = c1.DeliveryCityID
ORDER BY ci.CityName, Customer1, Customer2;
```

10) Orders that include “Chocolate” items

Proposition: Show orders that contain products with “Chocolate” in the name.

Question: Which orders include any chocolate item and how many units?

Answer: Join orders → lines → items and filter by item name.

How I’m solving it: Orders → OrderLines → StockItems; WHERE StockItemName LIKE '%Chocolate%'; sort by date.

Why special: A three-table join plus a text search on a joined table.

```python
SELECT
  o.OrderID, o.OrderDate,
  si.StockItemName,
  ol.Quantity
FROM Sales.Orders AS o
JOIN Sales.OrderLines AS ol ON ol.OrderID = o.OrderID
JOIN Warehouse.StockItems AS si ON si.StockItemID = ol.StockItemID
WHERE si.StockItemName LIKE '%Chocolate%'
ORDER BY o.OrderDate DESC, si.StockItemName;
```
