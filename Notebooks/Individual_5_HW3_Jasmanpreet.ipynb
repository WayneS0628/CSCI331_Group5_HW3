2. Question:
Which delivery methods could be used, and how many customers are set up for each?

Count customers by their preferred delivery method and show a list from most to least popular.
4. How I'm solving it:
Join Sales.Customers → Application.DeliveryMethods, count customers per method, group by DeliveryMethodName, and order the results from most to least.
5. Why it's special:
This demonstrates multi-table joins, aggregation (COUNT), and grouping (GROUP BY), which are all fundamental and frequently used SQL skills.

```python
SELECT
    dm.DeliveryMethodName,
    COUNT(c.CustomerID) AS NumberOfCustomers
FROM
    Sales.Customers AS c
JOIN
    Application.DeliveryMethods AS dm
    ON c.DeliveryMethodID = dm.DeliveryMethodID
GROUP BY
    dm.DeliveryMethodName
ORDER BY
    NumberOfCustomers DESC;
```

2. Question:
Which salespeople have the most orders?
3. Answer:

4. How I'm solving it:
Join the Sales.Orders table with the Application.People table, count the number of orders, group the results by the salesperson's name, and then sort the list from most orders to least.
5. Why it's special:
This demonstrates how to create a leaderboard by combining a JOIN, an aggregation (COUNT), and a grouping (GROUP BY).

```python
SELECT TOP 10
    p.FullName AS Salesperson,
    COUNT(o.OrderID) AS NumberOfOrders
FROM
    Sales.Orders AS o
JOIN
    Application.People AS p
    ON o.SalespersonPersonID = p.PersonID
GROUP BY
    p.FullName
ORDER BY
    NumberOfOrders DESC;
```

1. Proposition Six:
